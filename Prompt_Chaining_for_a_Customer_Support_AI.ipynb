# Exercise 1: Prompt Chaining for a Customer Support AI

**Goal:** Build a simple multi-step customer service flow where the output of one prompt becomes the input for the next prompt.

**Tools Used:**
- Google Colab
- Python
- ChatGPT

## Prompt Chain Design

**Customer Message:**  
"I ordered a pair of shoes five days ago, but my package still hasn't arrived. Can you help me?"

**Step 1 – Classify the Issue**  
Prompt: "You are a customer support assistant. Classify the customer's issue into one category: Shipping, Billing, Returns, Product Issue, or Other. Return only the category."

**Step 2 – Identify Missing Information**  
Prompt: "Based on the customer message and issue category, identify any information needed to help solve the problem. Ask only for necessary information and do not ask for sensitive information."

**Step 3 – Propose a Solution**  
Prompt: "Using the customer message, issue category, and information identified in the previous step, propose a short and helpful solution. Use a professional and friendly tone."

**Step 4 – Escalation Decision**  
Prompt: "Review the customer issue and proposed solution. Decide whether the issue should be escalated to a human support agent. Return Yes or No and give one short reason."

```python
# Customer message
customer_message = """
I ordered a pair of shoes five days ago, but my package still hasn't arrived.
Can you help me?
"""

print("Customer Message:")
print(customer_message)
```

```python
# Step 1: Classify the Issue

def classify_issue(customer_message):
    message = customer_message.lower()

    if "package" in message or "arrived" in message or "shipping" in message:
        return "Shipping"
    elif "bill" in message or "payment" in message or "charged" in message:
        return "Billing"
    elif "return" in message or "refund" in message:
        return "Returns"
    elif "broken" in message or "damaged" in message:
        return "Product Issue"
    else:
        return "Other"

issue_category = classify_issue(customer_message)

print("Step 1 - Issue Category:")
print(issue_category)
```

```python
# Step 2: Identify Missing Information
# This step uses the output from Step 1 (issue_category)

def identify_missing_info(customer_message, issue_category):
    if issue_category == "Shipping":
        return "Please provide your order number so I can check the shipping status."
    elif issue_category == "Billing":
        return "Please provide your order number and a brief description of the billing issue."
    elif issue_category == "Returns":
        return "Please provide your order number and the reason for the return."
    elif issue_category == "Product Issue":
        return "Please provide your order number and a description of the product problem."
    else:
        return "Please provide more details about the issue."

missing_info = identify_missing_info(customer_message, issue_category)

print("Step 2 - Missing Information:")
print(missing_info)
```

```python
# Step 3: Propose a Solution
# This step uses outputs from Step 1 and Step 2

def propose_solution(customer_message, issue_category, missing_info):
    if issue_category == "Shipping":
        return (
            "I’m sorry your package has not arrived yet. "
            "Once you provide your order number, we can check the tracking "
            "information and shipping status for you."
        )
    elif issue_category == "Billing":
        return (
            "We can review the billing issue after receiving your order number "
            "and details about the charge."
        )
    elif issue_category == "Returns":
        return (
            "We can help with your return after receiving your order number "
            "and reason for the return."
        )
    elif issue_category == "Product Issue":
        return (
            "We can review the product issue and determine the appropriate "
            "replacement or support option."
        )
    else:
        return "Please provide more information so we can help resolve your issue."

solution = propose_solution(customer_message, issue_category, missing_info)

print("Step 3 - Proposed Solution:")
print(solution)
```

```python
# Step 4: Escalation Decision
# This step uses the customer issue and proposed solution

def escalation_decision(issue_category, solution):
    if issue_category == "Other":
        return "Yes - The issue is unclear and should be reviewed by a human support agent."
    else:
        return "No - A clear solution or next step has been provided to the customer."

escalation = escalation_decision(issue_category, solution)

print("Step 4 - Escalation Decision:")
print(escalation)
```

```python
# Testing the prompt chain with a second customer message

test_message = """
I received my shoes, but they are badly damaged and I need help.
"""

test_category = classify_issue(test_message)
test_missing_info = identify_missing_info(test_message, test_category)
test_solution = propose_solution(test_message, test_category, test_missing_info)
test_escalation = escalation_decision(test_category, test_solution)

print("Test Customer Message:")
print(test_message)

print("Step 1 - Issue Category:")
print(test_category)

print("Step 2 - Missing Information:")
print(test_missing_info)

print("Step 3 - Proposed Solution:")
print(test_solution)

print("Step 4 - Escalation Decision:")
print(test_escalation)
```

## Testing and Iteration

The first test correctly classified the damaged shoes as a Product Issue. However, the escalation step returned "No" even though a badly damaged product may require assistance from a human support agent.

To improve the prompt chain, I refined the escalation rule so that serious product issues are escalated to a human agent.

```python
# Improved Step 4: Escalation Decision
# Product issues are now escalated to a human support agent

def improved_escalation_decision(issue_category, solution):
    if issue_category == "Product Issue":
        return "Yes - A damaged product may require replacement or additional help from a human support agent."
    elif issue_category == "Other":
        return "Yes - The issue is unclear and should be reviewed by a human support agent."
    else:
        return "No - A clear solution or next step has been provided to the customer."

improved_escalation = improved_escalation_decision(
    test_category, test_solution
)

print("Improved Step 4 - Escalation Decision:")
print(improved_escalation)
```

## Final Prompt Chain Summary

- **Step 1:** Classifies the customer's message into an issue category.
- **Step 2:** Uses the customer message and Step 1 category to identify missing information.
- **Step 3:** Uses the customer message, issue category, and missing information to propose a solution.
- **Step 4:** Uses the issue category and proposed solution to determine whether human escalation is needed.
- **Iteration:** Testing showed that a damaged product was not initially escalated. The escalation rule was improved so serious product issues are sent to a human support agent.
