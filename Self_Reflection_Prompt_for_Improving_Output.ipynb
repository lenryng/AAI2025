# Exercise 3: Self-Reflection Prompt for Improving Output

**Goal:** Use a self-reflection prompt to ask the AI to critique and improve its own summary based on specific requirements.

**Tools Used:**
- Google Colab
- Python
- ChatGPT

## Original Summary

**Topic: Artificial Intelligence in Business**

Artificial intelligence is becoming important in business. Companies use AI to analyze data, automate tasks, and help customers. AI can help businesses save time and make better decisions. However, companies also need to consider problems such as bias, privacy, and the responsible use of AI.

## Self-Reflection Prompt

Review the original summary about artificial intelligence in business.

Critique the summary using the following criteria:

1. **Accuracy:** Check that the information is accurate and does not make unsupported claims.
2. **Clarity:** Make sure the ideas are easy to understand and logically organized.
3. **Audience:** Write for a college-level student who is learning about AI in business.
4. **Length:** Keep the improved summary between 100 and 150 words.
5. **Formatting:** Use one clear paragraph with complete sentences.
6. **Focus:** Explain both the benefits and challenges of using AI in business.

First, briefly identify the weaknesses in the original summary. Then provide a revised version that addresses those weaknesses.

## Self-Critique

The original summary is accurate and easy to understand, but it is too general and brief. It mentions several benefits of AI but does not explain them in much detail. The challenges of bias, privacy, and responsible AI use are also mentioned without explaining why they matter. The summary could be improved by providing clearer connections between AI applications and business benefits while still discussing potential risks.

## Improved Summary

Artificial intelligence is increasingly used by businesses to improve efficiency, analyze data, automate repetitive tasks, and support customer service. AI can help organizations process large amounts of information more quickly, identify useful patterns, and support better business decisions. Automation can also reduce the time employees spend on routine work, allowing them to focus on more complex tasks. However, adopting AI also creates challenges. Businesses must consider data privacy, potential bias in AI systems, and the need for responsible human oversight. Poor-quality data or biased systems can lead to unreliable or unfair results. Therefore, businesses should balance the benefits of AI with appropriate policies, monitoring, and human involvement to use the technology effectively and responsibly.

## Reflection Results

After applying the self-reflection prompt, the revised summary improved the original by:

- Adding more specific explanations of AI benefits.
- Explaining the importance of privacy and bias.
- Including the role of human oversight.
- Improving clarity and organization.
- Keeping the summary appropriate for a college-level audience.
- Keeping the revised summary within the 100–150 word requirement.

```python
# Verify the improved summary meets the length requirement

improved_summary = """
Artificial intelligence is increasingly used by businesses to improve efficiency, analyze data, automate repetitive tasks, and support customer service. AI can help organizations process large amounts of information more quickly, identify useful patterns, and support better business decisions. Automation can also reduce the time employees spend on routine work, allowing them to focus on more complex tasks. However, adopting AI also creates challenges. Businesses must consider data privacy, potential bias in AI systems, and the need for responsible human oversight. Poor-quality data or biased systems can lead to unreliable or unfair results. Therefore, businesses should balance the benefits of AI with appropriate policies, monitoring, and human involvement to use the technology effectively and responsibly.
"""

word_count = len(improved_summary.split())

print("Improved Summary Word Count:", word_count)

if 100 <= word_count <= 150:
    print("Result: The improved summary meets the 100-150 word requirement.")
else:
    print("Result: The summary needs to be revised to meet the word requirement.")
```
